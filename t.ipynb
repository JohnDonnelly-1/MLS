```python
import decimal
import math

def my_sum(val, iters):
    # 1. Determine the level of precision in the value that will be summed
    
    decimal_places = abs(decimal.Decimal(str(val)).as_tuple().exponent)
    bad_values = []
    for it in range(iters):
        total = math.fsum(val for _ in range(1,it))
        precision_of_total = abs(decimal.Decimal(str(total)).as_tuple().exponent)
        if decimal_places < precision_of_total:
            bad_values.append((val,it,total))
    return bad_values

def my_reg_sum(val, iters):
    # 1. Determine the level of precision in the value that will be summed
    
    decimal_places = abs(decimal.Decimal(str(val)).as_tuple().exponent)
    bad_values = []
    for it in range(iters):
        total = sum(val for _ in range(1,it))
        precision_of_total = abs(decimal.Decimal(str(total)).as_tuple().exponent)
        if decimal_places < precision_of_total:
            bad_values.append((val,it,total))
    return bad_values
    
    
bad = my_sum(.25,10000)
bad2 = my_reg_sum(.125,10000)
print(len(bad))
print(len(bad2))
```

```python
def fibonacci(n):
  if n <= 0:
    return []
  elif n == 1:
    return [0]
  elif n == 2:
    return [0, 1]
  else:
    fib_list = [0, 1]
    for i in range(2, n):
      fib_list.append(fib_list[i-1] + fib_list[i-2])
    return fib_list

terms = int(input("How many terms of the Fibonacci sequence do you want to generate? "))
print(fibonacci(terms))
```
